```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Dataset Sintetis Pasien
data = {
    'Suhu_Tubuh': [36.5, 38.5, 37.0, 39.0, 36.2, 38.8, 36.8, 39.5],
    'Detak_Jantung': [70, 110, 75, 115, 68, 105, 72, 120],
    'Status': [0, 1, 0, 1, 0, 1, 0, 1] # 0: Sehat, 1: Demam
}
df = pd.DataFrame(data)

# 2. Split Fitur (X) dan Target (y)
X = df[['Suhu_Tubuh', 'Detak_Jantung']]
y = df['Status']

# 3. Train-Test Split (75% Latih, 25% Uji)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 4. Inisialisasi & Fit
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# 5. Predict & Score
y_pred = knn.predict(X_test)
print(f"Akurasi Model: {accuracy_score(y_test, y_pred) * 100}%")
```

```python
pasien_baru = [[39.0, 115]]
prediksi_pasien = knn.predict(pasien_baru)

if prediksi_pasien[0] == 1:
    print("Pasien terdeteksi demam")
else:
    print("Pasien sehat")
```

```python
# Menguji nilai K dari 1 sampai 5
for k in range(1, 6, 2):  # Menguji K = 1, 3, 5
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    score = accuracy_score(y_test, model.predict(X_test))
    print(f"Akurasi untuk K={k} : {score * 100:.1f}%")
```
